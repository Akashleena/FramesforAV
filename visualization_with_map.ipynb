```python
import numpy as np
import matplotlib.pyplot as plt
import matplotlib.cm as cm
from nuscenes.nuscenes import NuScenes
from pyquaternion import Quaternion
from collections import defaultdict, OrderedDict
from nuscenes.map_expansion.map_api import NuScenesMap

# Initialize nuScenes dataset
nusc = NuScenes(version='v1.0-mini', dataroot='/Users/liliu/Desktop/Nuscenes/nuscenes', verbose=True)

# Select a scene (e.g., 2nd scene)
scene = nusc.scene[2]
first_sample_token = scene['first_sample_token']

# Load map for this scene
log = nusc.get("log", scene["log_token"])
location = log["location"]
nusc_map = NuScenesMap(dataroot='/Users/liliu/Desktop/Nuscenes/nuscenes', map_name=location)

# Initialize containers
ego_trajectory, ego_orientations, timestamps = [], [], []
object_trajectories, object_orientations, object_labels = defaultdict(list), defaultdict(list), {}
object_name_counter = defaultdict(int)

# Marker shapes by category root
category_marker_map = {
    'vehicle': 's',            # star
    'movable_object': '^',     # triangle
    'human.pedestrian': '*',   # square
    'animal': 'D',             # diamond
    'static_object': 'x',      # x mark
    'flat': 'P'                # plus
}

# Convert quaternion to yaw
def quaternion_to_yaw(quaternion):
    q = Quaternion(quaternion)
    return np.arctan2(2.0 * (q.w * q.z + q.x * q.y), 1.0 - 2.0 * (q.y ** 2 + q.z ** 2))

# Loop over all samples
sample_token = first_sample_token
while sample_token:
    sample = nusc.get('sample', sample_token)
    timestamps.append(sample['timestamp'])

    # Ego pose
    sd = nusc.get('sample_data', sample['data']['LIDAR_TOP'])
    pose = nusc.get('ego_pose', sd['ego_pose_token'])
    pos = tuple(pose['translation'][:2])
    yaw = quaternion_to_yaw(pose['rotation'])
    ego_trajectory.append(pos)
    ego_orientations.append(yaw)

    # Annotations
    for ann_token in sample['anns']:
        ann = nusc.get('sample_annotation', ann_token)
        inst_token = ann['instance_token']
        cat = ann['category_name']
        pos = tuple(ann['translation'][:2])
        yaw = quaternion_to_yaw(ann['rotation'])

        if inst_token not in object_labels:
            object_name_counter[cat] += 1
            label = f"{cat}_{object_name_counter[cat]}"
            object_labels[inst_token] = label

        object_trajectories[inst_token].append(pos)
        object_orientations[inst_token].append(yaw)

    sample_token = sample['next']

# Normalize time
timestamps = np.array(timestamps)
timestamps = (timestamps - timestamps.min()) / (timestamps.max() - timestamps.min())

ego_trajectory = np.array(ego_trajectory)
ego_orientations = np.array(ego_orientations)

# Filter unique locations per instance
unique_location_counts = {}
filtered_trajectories = {}
filtered_orientations = {}

for inst, traj in object_trajectories.items():
    seen = OrderedDict()
    for i, loc in enumerate(traj):
        if loc not in seen:
            seen[loc] = object_orientations[inst][i]
    unique_location_counts[inst] = len(seen)
    filtered_trajectories[inst] = list(seen.keys())
    filtered_orientations[inst] = list(seen.values())

# Top 10 by unique location length
sorted_lengths = sorted(set(unique_location_counts.values()), reverse=True)
selected = (
    {k for k, v in unique_location_counts.items() if v >= sorted_lengths[9]}
    if len(sorted_lengths) > 10 else set(unique_location_counts.keys())
)

# Generate color map
color_map = cm.get_cmap("tab10", len(selected))
instance_colors = {inst: color_map(i) for i, inst in enumerate(selected)}

# Plot base map
x_min, y_min = np.min(ego_trajectory, axis=0) - 50
x_max, y_max = np.max(ego_trajectory, axis=0) + 50
fig, ax = nusc_map.render_map_patch((x_min, y_min, x_max, y_max), ['drivable_area', 'walkway'], alpha=0.05)
ax.set_aspect('equal')

# Plot selected instances
legend_labels = {}

for inst in selected:
    traj = np.array(filtered_trajectories[inst])
    yaws = np.array(filtered_orientations[inst])
    name = object_labels[inst]
    color = instance_colors[inst]

    cat_root = name.rsplit('_', 1)[0].split('.')[0]
    marker = category_marker_map.get(cat_root, 'o')

    legend_labels[name] = (marker, color[:3], cat_root)  # include category for grouping

    for i in range(len(traj)):
        alpha = timestamps[i]
        ax.plot(traj[i, 0], traj[i, 1], marker=marker, color=(*color[:3], alpha),
                markersize=6, linestyle='None')
        ax.arrow(traj[i, 0], traj[i, 1], 0.5 * np.cos(yaws[i]), 0.5 * np.sin(yaws[i]),
                 head_width=0.3, head_length=0.3, fc=(*color[:3], alpha), ec=(*color[:3], alpha))

# Plot ego trajectory
for i in range(len(ego_trajectory)):
    alpha = timestamps[i]
    ax.plot(ego_trajectory[i, 0], ego_trajectory[i, 1], marker='o', color=(0, 0, 0, alpha))
    ax.arrow(ego_trajectory[i, 0], ego_trajectory[i, 1], 0.5 * np.cos(ego_orientations[i]),
             0.5 * np.sin(ego_orientations[i]), head_width=0.3, head_length=0.3,
             fc=(0, 0, 1, alpha), ec=(0, 0, 1, alpha))

# Group legend entries
grouped_legend = defaultdict(list)
for name, (marker, color, cat_root) in legend_labels.items():
    grouped_legend[cat_root].append((name, marker, color))

# Optional order
category_order = ['vehicle', 'movable_object', 'human']

# Construct grouped legend entries
legend_entries = []
for cat in category_order:
    if cat in grouped_legend:
        # Add a group title with dashed line
        # legend_entries.append(
        #     plt.Line2D([0], [0], linestyle='--', linewidth=1, color='black', label=f'[{cat}]')
        # )
        for name, marker, color in grouped_legend[cat]:
            entry = plt.Line2D([0], [0], marker=marker, color=color, markersize=8,
                               linestyle='None', label=name)
            legend_entries.append(entry)

plt.title(f"Ego Vehicle and Top 10 Unique Object Trajectories on nuScenes Map ({location})")
plt.xlabel("X (meters)")
plt.ylabel("Y (meters)")
plt.legend(handles=legend_entries, loc='upper right', fontsize=8, frameon=True)
plt.grid()
plt.show()
```
